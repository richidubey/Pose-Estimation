```python
import plotly.express as px
import cv2

IMAGE_PATH = "/coc/flash5/rdubey36/datasets/ForkliftScene3Dv2/ForkliftScene3Dv2/rgb/rgb_00000.png"

# Read image and convert from BGR (OpenCV default) to RGB
img = cv2.imread(IMAGE_PATH)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display interactive Plotly figure
fig = px.imshow(img, title="Hover over the pallet to get X, Y coordinates")
fig.update_layout(
    width=1200,   # increase width
    height=900    # increase height
)
fig.show()
```
